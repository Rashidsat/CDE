2. Write a program that reads a string from keyboard and display: 
    "* The number of uppercase letters in the string 
    "* The number of lowercase letters in the string 
    "* The number of digits in the string 
    "* The number of whitespace characters in the string 


```python
p = input('Enter the words plz')
upper = lower = digit= spaces = spex =  0
for text in p :
        if text.isupper():
            upper = upper+1
            
        elif text.islower():
            lower = lower+1
           
        elif text.isdigit():
            digit = digit+1
            
        elif text.isspace():
            spaces=spaces+1
        else:
            spex = spex + 1
            
print ('Uppercase =:', upper)
print ('lowercase =:', lower)
print ('digit=:', digit )
print ('spaces=:', spaces)
print ('others', spex)
```

3. Write a Python program that accepts a string from user. Your program should create and display a new string where the first and last characters have been exchanged. \n",
"##### For example if the user enters the string 'HELLO' then new string would be 'OELLH' "


```python
d = input("Enter a word: ")

r = d[-1] + d[1:-1] + d[0]

print("word:", r)
```
